```python
x = torch.tensor(3., requires_grad=True)
y = x ** 2

y.backward()
print(x.grad)
```

```python
x1 = torch.tensor(2., requires_grad=True)
x2 = torch.tensor(3., requires_grad=True)

w1 = torch.tensor(0.5, requires_grad=True)
w2 = torch.tensor(0.2, requires_grad=True)

b = torch.tensor(0.1, requires_grad=True)

y = w1 * x1 + w2 * x2 + b

y.backward()
```

```python
print(w1.grad)
print(w2.grad)
print(b.grad)
```

```python
import torch.nn as nn
```

```python
relu = nn.ReLU()
```

```python
X = torch.tensor([[2, 3]], dtype=torch.float32)   # one sample, two features
y = torch.tensor([[1]], dtype=torch.float32)      # actual class
W1 = torch.tensor([
    [0.5, 0.1],
    [0.2, 0.4]
], dtype=torch.float32, requires_grad=True)

b1 = torch.tensor([
    [0.1, 0.2]
], dtype=torch.float32, requires_grad=True)

Z1 = X @ W1 + b1
A1 = relu(Z1)

W2 = torch.tensor([
    [0.2],
    [0.3]
], dtype=torch.float32, requires_grad=True)

b2 = torch.tensor([
    [0.1]
], dtype=torch.float32, requires_grad=True)

Z2 = A1 @ W2 + b2
y_hat = torch.sigmoid(Z2)
```

```python
loss_fn = torch.nn.BCELoss()
loss = loss_fn(y_hat, y)
```

```python
loss.backward()
```

```python
print(w1.grad)
print(w2.grad)
print(b1.grad)
print(b2.grad)
```
