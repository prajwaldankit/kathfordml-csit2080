```python
class LogisticRregression:
    def __init__(self, learning_rate=0.001, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = 0
        self.loss = []

    @staticmethod
    def _cross_entropy_loss(y, y_hat):
        n = y.shape[0]
        epsilon = 1e-15
        loss = - np.sum(y * np.log(y_hat + epsilon) + (1-y) * np.log(1-y_hat + epsilon)) / n
        return loss

    def _sigmoid(self,X):
        z = self.weights * X + self.bias
        return 1 / (1 + np.exp(-z))


    def _gradient_descent(self, X, y):
        n = y.shape[0]
        y_hat = self._sigmoid(X)

        de_dw = (1 / n) * np.dot(X.T, (y_hat - y))  
        de_db = (1 / n) * np.sum(y_hat - y)

        self.weights = self.weights - self.learning_rate * de_dw
        self.bias = self.bias - self.learning_rate * de_db
      

    def fit(self, X, y):
        _, m = X.shape


        self.weights = np.zeros(m)
        self.bias = 0.0

        for _ in range(self.n_iterations):
            self._gradient_descent(X, y)
            y_hat = self._sigmoid(X)
            loss = self._cross_entropy_loss(y, y_hat)
            self.loss.append(loss)


    def predict(self, X):
        return self._sigmoid(X)
        

model = LogisticRregression()
        
```
